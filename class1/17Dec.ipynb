# Generator

```python
def my_generator():
	# simple generator function
	a = 1
	print("First Time")
	yield a

	a += 1
	print("Second Time")
	yield a

	a += 1
	print("Third Time")
	yield a

# create generator and retrieve values with next()
gen = my_generator()
print(next(gen))  # First Time -> 1
print(next(gen))  # Second Time -> 2
print(next(gen))  # Third Time -> 3
```
