```python
import pandas as pd
india_weather=pd.DataFrame({
    "city":["mumbai","delhi","banglore"],
    "temperature":[32,45,30],
    "humidity":[80,90,98]
})
india_weather
```

```python
us_weather=pd.DataFrame({
    "city":["New York","Chicago","orlando"],
    "Temperature":[22,33,24],
    "Humidity":[65,88,67]
})
us_weather
```

```python
#concate Teo dataframes
df=pd.concat([india_weather,us_weather])
df
```

```python
# if you want continous index
df=pd.concat([india_weather,us_weather],ignore_index=True,axis=0)
df
```

```python
# if you want continous index
df=pd.concat([india_weather,us_weather],ignore_index=True,axis=1)
df
```

# Merge DAtaFrames

```python
temperature_df=pd.DataFrame({
    "city":["mumbai","delhi","banglore","Hydrabad"],
    "temperature":[22,33,44,22]})
temperature_dtemperature_df=pd.DataFrame({
    'city':["mumbai","Delhi","banglore","hydrabad"],
    "termperature":[32,45,30,40]
})
temperature_df

```

```python
humidity_df=pd.DataFrame({
    "city":["delhi","mumbai","banglore"],
    "humidity":[22,32,23]
})
humidity_df
```

```python
#merge Two dataframe with out explicitly mention index
df=pd.merge(temperature_df,humidity_df,on='city')
df
```

```python
    # Outer Join
df=pd.merge(temperature_df,humidity_df,on='city',how ='outer')
df
```
